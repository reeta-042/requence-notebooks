```python
import pandas as pd

# 1. Load the matrices using the Tab separator (\t)
# This solves the problem of the headers being stuck together with \t
merged = pd.read_csv('Downloads/E_coli/e_coli_genome/master_gene_matrix.csv', sep='\t')

# 2. Clean and Rename the ID column
    # ABRicate uses '#FILE' as the first column header
merged['#FILE'] = merged['#FILE'].str.replace('.fna', '', regex=False).str.replace('.tab', '', regex=False)
merged.rename(columns={'#FILE': 'Genome_ID'}, inplace=True)


# 4. Convert to Binary (1s and 0s)
merged.replace('.', 0, inplace=True)
merged.fillna(0, inplace=True)

def make_binary(val):
    val_str = str(val).strip()
    # Check for empty, zero, or dot values
    if val_str in ['0', '', '0.0', '.', 'nan']:
        return 0
    return 1

# Apply to all columns except identifiers
cols_to_binarize = [c for c in merged.columns if c not in ['Genome_ID', 'NUM_FOUND']]
for col in cols_to_binarize:
    merged[col] = merged[col].apply(make_binary)

# 5. Save the Professional Master Table
merged = merged.drop(columns = ['NUM_FOUND'])
merged.to_csv('Downloads/E_coli/master_gene_matrix_coli.csv', index=False)

print(f"Merge Successful!")
print(f"Total Isolates: {merged.shape[0]}")
print(f"Total AMR Features: {len(cols_to_binarize)}")
```

```python
import pandas as pd

# 1. Load the matrices using the Tab separator (\t)
# This solves the problem of the headers being stuck together with \t
card = pd.read_csv('Downloads/E_coli/e_coli_genome/gene_presence_matrix.csv', sep='\t')
res = pd.read_csv('Downloads/E_coli/e_coli_genome/resfinder_presence_matrix.csv', sep='\t')

# 2. Clean and Rename the ID column
for df in [card, res]:
    # ABRicate uses '#FILE' as the first column header
    df['#FILE'] = df['#FILE'].str.replace('.fna', '', regex=False).str.replace('.tab', '', regex=False)
    df.rename(columns={'#FILE': 'Genome_ID'}, inplace=True)

# 3. Horizontal Merge (Outer Join)
# This ensures each Genome_ID has one row with both CARD and ResFinder data
merged = pd.merge(card, res, on='Genome_ID', how='outer', suffixes=('_card', '_res'))

# 4. Convert to Binary (1s and 0s)
merged.replace('.', 0, inplace=True)
merged.fillna(0, inplace=True)

def make_binary(val):
    val_str = str(val).strip()
    # Check for empty, zero, or dot values
    if val_str in ['0', '', '0.0', '.', 'nan']:
        return 0
    return 1

# Apply to all columns except identifiers
cols_to_binarize = [c for c in merged.columns if c not in ['Genome_ID', 'NUM_FOUND_card', 'NUM_FOUND_res']]
for col in cols_to_binarize:
    merged[col] = merged[col].apply(make_binary)

# 5. Save the Professional Master Table
merged = merged.drop(columns = ['NUM_FOUND_card','NUM_FOUND_res'])
merged.to_csv('Downloads/E_coli/e_coli_genome/gene_features.csv', index=False)

print(f"Merge Successful!")
print(f"Total Isolates: {merged.shape[0]}")
print(f"Total AMR Features: {len(cols_to_binarize)}")
```
